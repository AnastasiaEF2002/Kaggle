```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Loading the Data
mydata= pd.read_csv('/kaggle/input/ai-index/AI_index_db.csv',encoding='ISO-8859-1')
df = pd.DataFrame(mydata)
df=df.dropna()


# Data Insights
print("===================================# VIEWING AND INSPECTING DATA =================================== \n")
print("---------- # FIRST 5 ROWS # -----------\n",df.head())
print("---------- #  LAST 5 ROWS# # -----------\n",df.tail())
print("---------- #  SUMMARY OF THE DATAFRAME# # -----------\n",df.info())
print("---------- #  STATISTICS OF NUMERICAL COLUMNS# # -----------\n",df.describe())
print("---------- #  SHAPE OF THE DATAFRAME# # -----------\n",df.shape)
print("---------- #  LIST OF COLUMN NAMES# # -----------\n",df.columns)
print("---------- #  DATA TYPES OF COLUMNS# # -----------\n",df.dtypes)

# Data Handling
print("===================================# DATA HANDLING # =================================== \n")
print("---------- # NULL VALUE INFO # -----------\n",df.isnull().sum())

# Selecting only x and y for clustering
X = df[['Talent', 'Infrastructure']]

# Applying K-Means clustering (choosing 3 clusters as an assumption)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X)

# Plotting the clusters
plt.figure(figsize=(8, 6))
plt.scatter(df['Talent'], df['Infrastructure'], c=df['cluster'], cmap='viridis', alpha=0.6, edgecolors='k')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='red', marker='X', label='Centroids')

plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.title('K-Means Clustering of Data')
plt.legend()
plt.show()
```
